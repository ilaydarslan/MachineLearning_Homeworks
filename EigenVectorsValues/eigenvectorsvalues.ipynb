# Eigenvalues and Eigenvectors Calculation

Bu çalışmada özdeğerler manuel olarak hesaplanmış ve NumPy kütüphanesi ile karşılaştırılmıştır.

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

print("Matris A:")
print(A)
```

Özdeğerler karakteristik polinom kullanılarak hesaplanır.

det(A - λI) = 0

```python
# Karakteristik polinom
coefficients = np.poly(A)

# Manuel özdeğerler
eigenvalues_manual = np.roots(coefficients)

print("Manuel hesaplanan özdeğerler:")
print(eigenvalues_manual)
```

```python
eigenvalues_np, eigenvectors_np = np.linalg.eig(A)

print("NumPy özdeğerler:")
print(eigenvalues_np)

print("NumPy özvektörler:")
print(eigenvectors_np)
```
